```python
# Cell 3: Load the data
import pandas as pd

movies = pd.read_csv('tmdb_5000_movies.csv')
credits = pd.read_csv('tmdb_5000_credits.csv')

print(movies.shape)
movies.head()
```

To add a new movie, you need to create a new row (as a dictionary or pandas Series) with the movie's information, ensuring the column names match your existing `movies` DataFrame. Then, you can append this new row to the DataFrame.

```python
# Create a new movie entry for 'jananayagan'
new_movie_data = {
    'budget': 0, # Assuming budget is unknown or 0 for now
    'genres': '[{"id": 99, "name": "Documentary"}]', # Example genre
    'homepage': '',
    'id': 999999, # Assign a unique ID for the new movie
    'keywords': '[]',
    'original_language': 'ta', # Assuming Tamil
    'original_title': 'Jananayagan',
    'overview': 'A movie about a leader.',
    'popularity': 0.0,
    'production_companies': '[]',
    'production_countries': '[]',
    'release_date': '2023-10-27',
    'revenue': 0,
    'runtime': 120.0,
    'spoken_languages': '[{"iso_639_1": "ta", "name": "Tamil"}]',
    'status': 'Released',
    'tagline': 'The Leader',
    'title': 'Jananayagan',
    'vote_average': 0.0,
    'vote_count': 0
}

# Convert the dictionary to a DataFrame row
new_movie_df = pd.DataFrame([new_movie_data])

# Append the new movie to the existing movies DataFrame
# Using ignore_index=True to re-index the DataFrame
movies = pd.concat([movies, new_movie_df], ignore_index=True)

# Display the last few rows to show the added movie
print("Movies DataFrame after adding 'Jananayagan':")
display(movies.tail())
```
